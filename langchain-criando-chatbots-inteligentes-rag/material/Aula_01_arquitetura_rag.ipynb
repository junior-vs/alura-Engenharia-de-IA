## O que vamos aprender:
### Como o RAG resolve problemas reais dos LLMs

O RAG (Retrieve, Augment, Generation) resolve problemas dos LLMs de algumas formas importantes:
Atualização de dados: O RAG permite que o LLM acesse informações mais recentes e relevantes, o que é crucial em um mundo onde os dados mudam constantemente.
Respostas verificadas: Ao buscar informações em fontes externas confiáveis, o RAG ajuda a garantir que as respostas do LLM sejam mais precisas e baseadas em evidências.
Custos otimizados: O RAG pode reduzir os custos associados ao treinamento contínuo de LLMs, pois permite que eles acessem informações externas em vez de depender apenas de seus dados de treinamento.
Aplicações reais importantes: O RAG possibilita o uso de LLMs em uma variedade de aplicações práticas, como chatbots de atendimento ao cliente, sistemas de recomendação e ferramentas de pesquisa.

Em resumo, o RAG aprimora a capacidade dos LLMs de fornecer informações precisas, atualizadas e relevantes, tornando-os mais úteis e confiáveis em uma variedade de contextos.

### Diferenças práticas entre RAG e prompting tradicional
A principal diferença prática entre RAG e prompting tradicional reside na capacidade de acessar e utilizar informações externas para gerar respostas mais precisas e contextuais.No prompting tradicional, o LLM depende exclusivamente do conhecimento que adquiriu durante o treinamento. Isso significa que ele pode ter dificuldades em responder a perguntas sobre tópicos novos ou específicos, ou pode fornecer informações desatualizadas ou imprecisas.Já o RAG, como vimos, permite que o LLM busque informações relevantes em fontes externas e as incorpore em suas respostas. Isso o torna mais flexível, adaptável e capaz de lidar com uma gama maior de perguntas e cenários.No contexto da aula, vimos um exemplo prático dessa diferença. Ao perguntar ao LLM sobre a política de home office da empresa usando o prompting tradicional, ele respondeu que não tinha essa informação, pois não era uma empresa. No entanto, com o RAG, podemos fornecer um documento com a política de home office e obter uma resposta precisa e relevante.Em resumo, o RAG oferece uma abordagem mais robusta e eficaz para resolver problemas reais com LLMs, permitindo que eles acessem e utilizem informações externas para gerar respostas mais precisas, contextuais e atualizadas.

**Analogias para entender a aula sobre RAG:**

LLM como um estudante sem acesso à internet:
Imagine um estudante (o LLM) que fez todas as aulas e possui muito conhecimento, mas não tem acesso à internet (informações externas).
Quando você faz uma pergunta sobre um tópico específico, ele tenta responder com o que aprendeu, mas pode não ter a informação mais recente ou precisa.

Essas analogias ilustram como o RAG aprimora a capacidade do LLM de acessar e utilizar informações externas, tornando-o mais flexível, adaptável e capaz de lidar com uma variedade maior de perguntas e cenários.

## 1. Prompting Tradicional vs RAG - Comparação Prática

Vamos ver na prática a diferença entre usar apenas um LLM (prompting tradicional) e usar RAG.

````python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,
    # other params...
)
````

````python
# EXEMPLO 1: Prompting Tradicional (sem RAG)

pergunta = "Qual é a política de home office da nossa empresa?"

prompt_tradicional = ChatPromptTemplate.from_template(
    "Responda a seguinte pergunta: {pergunta}"
)
````

````python
chain_tradicional = prompt_tradicional | llm

resposta_tradicional = chain_tradicional.invoke({"pergunta": pergunta})
````

````python
print(resposta_tradicional.content)
````

-----
## OBS: O que é esse | ?

O operador `|` em Python, nesse contexto com **LangChain**, é uma **forma elegante de compor etapas de uma cadeia (chain)** de execução, como se fosse um **"pipeline"**.

### ✅ Significado prático no LangChain:

```python
chain = prompt | llm
```

Esse código cria uma **cadeia (chain)** onde:

* O `prompt` é executado primeiro,
* E o resultado (texto formatado com as variáveis) é enviado diretamente para o `llm` (modelo de linguagem, como o Gemini),
* Retornando a **resposta gerada**.

## Carregando nosso Documento
Agora, em vez de simular o documento, vamos carregar o `politica_home_office.pdf` que está na mesma pasta.

````python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("/content/politica_home_office.pdf")

documento = loader.load()
````

````python
documento
````

````python
contexto_empresa = documento[0].page_content
````

````python
contexto_empresa
````

````python
#prints only its first 500 characters, followed by an ellipsis to indicate that there's more content.
#This helps to quickly verify that the document loaded correctly.
print(contexto_empresa[:500] + "...")
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, # Tamanho máximo de cada chunk
    chunk_overlap=200, # Quantos caracteres os chunks se sobrepõem (para contexto)
    length_function=len,
    add_start_index=True,
)

chunks_do_documento = text_splitter.split_documents(documento)

print(f"Documento original dividido em {len(chunks_do_documento)} chunks.")
print("Primeiro chunk:")
print(chunks_do_documento[0].page_content[:300] + "...")
print("Segundo chunk:")
print(chunks_do_documento[1].page_content[:300] + "...")
````

````python
# Exemplo 2: Com RAG - Usando o contexto do PDF

prompt_rag = ChatPromptTemplate.from_template("""
Use o contexto abaixo para responder a pergunta.
Se não souber a resposta baseado no contexto, diga que não tem a informação.

Contexto: {contexto}
Pergunta: {pergunta}

RespostaResposta:""")
````

````python
chain_rag = prompt_rag | llm

resposta_rag = chain_rag.invoke({"contexto": chunks_do_documento, "pergunta": pergunta})
````

````python
print(resposta_rag.content)
````
